```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
d=load_iris()
```

```python
print(d.DESCR)
```

```python
import seaborn as sns
df=sns.load_dataset("iris")
```

```python
df.head(5)
```

```python
d.target # in dataset=d species which is categorical feature is converted into numeric| type variable
```

```python
## INDEPENDENT AND DEPENDENT FEATURES
x=df.iloc[:,:-1]
y=d.target
```

```python

x,y
```

```python
##### TRAIN TEST SPLIT

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
## applying desicion treeee
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
##Post pruning
treeclassifier=DecisionTreeClassifier()
```

```python
treeclassifier.fit(x_train,y_train)
```

```python
from sklearn import tree
plt.figure(figsize=(15,10))
#tree.plot_tree(treeclassifier,filled=True)#IT WILL SHOW THE WHOLE TREE
```

```python
x_train.head()
```

```python
#POST PRUNING with max_depth=2
treeclassifier=DecisionTreeClassifier(max_depth=2)
treeclassifier.fit(x_train,y_train)
tree.plot_tree(treeclassifier,filled=True)
```

```python
## PREDICTION
y_pred=treeclassifier.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
score=accuracy_score(y_pred,y_test)
score
```

```python
c_report=classification_report(y_pred,y_test)
print(c_report)
```
