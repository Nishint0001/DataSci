```python
## WORKING ON BREAST CANCER DATASET
## USE PRE_PRUNING FOR LARGE DATASETS
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
from sklearn.datasets import load_breast_cancer
df=load_breast_cancer()
print(df)
```

```python
print(df.DESCR)
```

```python
# INDEPENDENT AND DEPENDENT VARIABLE
x=df.data
y=df.target
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
### HYPER_PARARAM_TUNING
```

```python
parameter={
    "criterion":["ginni","entrophy","log_loss"],
    "splitter":["best","random"],
    "max_depth":[1,2,3,4,5,None],
    "max_features":["auto","sqrt","log2"],
      "min_samples_split":[2, 5, 10]

}
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier
```

```python
treeclassifier=DecisionTreeClassifier()
clf=GridSearchCV(treeclassifier,param_grid=parameter,cv=5,scoring="accuracy")
```

```python
##TRAIN THE DATA
clf.fit(x_train,y_train)
```

```python
clf.best_params_
```

```python
y_pred=clf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
score=accuracy_score(y_pred,y_test)
score
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
c_report=classification_report(y_pred,y_test)
print(c_report)
```
